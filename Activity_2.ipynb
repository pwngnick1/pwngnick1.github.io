```python
## 1. Plot One Function
x = np.linspace(-10, 10, 100)

y = x**2

plt.plot(x, y)

plt.title("Plot of y = x²")
plt.xlabel("x")
plt.ylabel("y")

plt.grid()
plt.show()
```

```python
## 2. Plot Two Functions
x = np.linspace(-10, 10, 100)

y1 = x**2
y2 = x**2 + 10

plt.plot(x, y1, label="y = x²")
plt.plot(x, y2, label="y = x² + 10")

plt.title("Two Functions")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid()
plt.show()
```

```python
## 3. Read Chromosome Base Pair Data
df = pd.read_excel("chrosome_basepairs.xlsx")

df
```

```python
## 4. Calculate the Length of the Whole Genome
total_basepairs = df["basepais(bp)"].sum()

print("Total number of base pairs:", total_basepairs)
angstrom_per_basepair = 3.4

total_angstroms = total_basepairs * angstrom_per_basepair

print("Total genome length in Angstroms:", total_angstroms)
total_meters = total_angstroms * 1e-10

print("Total genome length in meters:", total_meters)
```

```python
## 5. Cost to Sequence the Whole Genome
genome_mbp = total_basepairs / 1_000_000

print("Genome size in million base pairs:", genome_mbp)
years = [2001, 2010, 2011, 2020]

cost_per_mbp = [10000, 1, 0.10, 0.01]
total_costs = []

for cost in cost_per_mbp:
    total_cost = genome_mbp * cost
    total_costs.append(total_cost)

for year, cost in zip(years, total_costs):
    print(year, f"${cost:,.2f}")
```

```python
## 6. Change in Whole Genome Sequencing Cost
plt.plot(years, total_costs, marker="o")

plt.title("Cost to Sequence the Whole Genome Over Time")
plt.xlabel("Year")
plt.ylabel("Total Cost ($)")

plt.grid()
plt.show()
```
