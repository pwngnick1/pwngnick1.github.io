```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = "/content/gtex_integrin_7_organs.xlsx"

df = pd.read_excel(file_path)

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nOrgans in the dataset:")
print(df["primary_site"].value_counts())
```

## What are Integrins?

Integrins are proteins found on the surface of cells. They help cells attach to other cells and to the extracellular matrix. Integrins also help with cell signaling, movement, and communication. In this activity, integrin gene expression will be compared between Lung and Breast tissue.

```python
lung = df[df["primary_site"] == "Lung"]
breast = df[df["primary_site"] == "Breast"]

print("Lung samples:", len(lung))
print("Breast samples:", len(breast))
```

```python
gene_columns = [col for col in df.columns if col.startswith("ITG")]

print("Integrin genes:")
print(gene_columns)
print("\nNumber of genes:", len(gene_columns))
```

```python
plt.figure(figsize=(15, 6))

plt.violinplot(
    [lung[gene].dropna() for gene in gene_columns],
    showmeans=True
)

plt.xticks(
    range(1, len(gene_columns) + 1),
    gene_columns,
    rotation=90
)

plt.title("Integrin Gene Expression in Lung")
plt.xlabel("Integrin Genes")
plt.ylabel("Gene Expression")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 6))

plt.violinplot(
    [breast[gene].dropna() for gene in gene_columns],
    showmeans=True
)

plt.xticks(
    range(1, len(gene_columns) + 1),
    gene_columns,
    rotation=90
)

plt.title("Integrin Gene Expression in Breast")
plt.xlabel("Integrin Genes")
plt.ylabel("Gene Expression")

plt.tight_layout()
plt.show()
```

```python
lung_breast = df[df["primary_site"].isin(["Lung", "Breast"])].copy()

print(lung_breast["primary_site"].value_counts())
```

```python
import seaborn as sns
```

```python
plot_data = lung_breast.melt(
    id_vars="primary_site",
    value_vars=gene_columns,
    var_name="Gene",
    value_name="Expression"
)

plot_data.head()
```

```python
plt.figure(figsize=(16, 7))

sns.violinplot(
    data=plot_data,
    x="Gene",
    y="Expression",
    hue="primary_site",
    split=True,
    inner="quart"
)

plt.title("Integrin Gene Expression: Lung vs Breast")
plt.xlabel("Integrin Genes")
plt.ylabel("Gene Expression")
plt.xticks(rotation=90)
plt.legend(title="Organ")

plt.tight_layout()
plt.show()
```

```python
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

X = df[gene_columns].copy()

# Fill any missing values with the mean of that gene
X = X.fillna(X.mean())

# Standardize gene expression values
X_scaled = StandardScaler().fit_transform(X)

print("Data ready for t-SNE:", X_scaled.shape)
```

```python
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30
)

tsne_results = tsne.fit_transform(X_scaled)

print("t-SNE complete!")
```

```python
tsne_df = pd.DataFrame({
    "tSNE1": tsne_results[:, 0],
    "tSNE2": tsne_results[:, 1],
    "Organ": df["primary_site"].values
})

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=tsne_df,
    x="tSNE1",
    y="tSNE2",
    hue="Organ",
    s=50,
    alpha=0.7
)

plt.title("t-SNE of Integrin Gene Expression by Organ")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.legend(title="Organ", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

## Results and Conclusion

The violin plots show the distribution of integrin gene expression in Lung and Breast tissues. The split violin plot makes it easier to compare the expression of each integrin gene between the two tissues and shows that some genes have different expression patterns between Lung and Breast samples.

The t-SNE plot uses integrin gene expression from all seven organ sites in the dataset. Samples that have similar integrin expression patterns appear closer together on the plot, while samples with different expression patterns appear farther apart. The clustering suggests that integrin gene expression varies between different tissue types.
